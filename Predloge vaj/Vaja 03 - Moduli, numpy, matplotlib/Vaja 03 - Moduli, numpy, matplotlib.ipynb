## Modul ``matplotlib``

```python
import matplotlib.pyplot as plt
```

#### Enostaven primer:

*Pomembno:* `matplotlib` na graf izrisuje točke - pare `(x, y)`! 

Pripraviti moramo numerični *polji točk* na abscisi in ordinati.

```python
x = np.linspace(0, 10, 100)
```

```python
plt.plot(x, np.sin(x**1.2), 'k.')
```
