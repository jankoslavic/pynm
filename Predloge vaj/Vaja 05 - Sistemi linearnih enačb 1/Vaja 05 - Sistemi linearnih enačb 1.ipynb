\begin{bmatrix}

### Enoličnost rešitve

```python
def ima_rešitev(A, b):
    """ Na podlagi A in b izpiše, ali je sistem enačb Ax=b rešljiv. """
    D = np.linalg.det(A)
    r_A = np.linalg.matrix_rank(A)
    r_Ab = np.linalg.matrix_rank(np.column_stack((A, b)))
    print(f'\tdet: {D}\n\trank(A): {r_A}\n\trank(Ab): {r_Ab}')
    if not D:
        if r_A != r_Ab: 
            print('Ni rešitve.')
        else: 
            print('Neskončno rešitev.')
    else:
        print('Sistem je rešljiv.')
```

$$\textrm{cond}(||\textbf{A}||)=||\textbf{A}||\,||\textbf{A}^{-1}||$$

**Vprašanje 8:** Pripravite funkcijo ``gauss_priprava``, ki kot argumenta sprejme matriko $\mathbf{A}$ in vektor $\mathbf{b}$, pripravi razširjeno matriko koeficientov $\mathbf{[A|b]}$ ter za vsako pivotno vrstico ``i``, $0 \leq i \leq n-1$, izračuna in izpiše konstanto $\lambda$ za vse vrstice pod pivotno ($i+1 \leq j \leq n$) ($n$ je število vrstic $\mathbf{A}$).
$$ \lambda = \frac{a_{ji}}{a_{ii}}$$

**Vprašanje 9:** Pripravite novo funkcijo ``gauss_eliminacija``, ki, če je koeficient $a_{ii}$ različen od 0, opravi zamenjavo:

$$E_j \leftarrow E_j - \lambda E_i$$

In vrne rezultat eliminacije. Ne pišite celotne kode, uporabite kodo obstoječe funkcije ``gauss_priprava``!

```python
def gauss_eliminacija(A, b):
    """ Z Gaussovo elimnacijo določi zgornjerikotno matriko za reševanje SLE. """
    Ab = np.column_stack((A, b))
    for i, pivot in enumerate(Ab[:-1]):
        for vrstica in Ab[i+1:]:
            m = vrstica[i] / pivot[i]
            vrstica -= m*pivot
    return Ab
```

**Vprašanje 10:** Pripravite funkcijo ``gauss_resitev``, ki na podlagi rezultata funkcije ``gauss_eliminacija`` izračuna vrednosti elementov vektorja neznank $\mathbf{x}$ z uporabo obratnega vstavljanja. Namig: Vrstni red vrstic matrike lahko obrnete z ``A[::-1]``.

**Obratno vstavljanje:** 
    Sistem z zgornje trikotno razširjeno matriko koeficientov rešujemo "od spodaj navzgor":
    
   $$x_i = \frac{b_i - \sum_{j=i+1}^{n}a_{ij} x_j}{a_{ii}}$$

```python
def gauss_rešitev(Ab):
    """ 
	Obratno vstavljanje za rešitev eliminiranega sistema. 
	V razmislek: drugačna koda kot na predavnjih, rezultat pa je isti!
	"""
    n = Ab.shape[0]
    x = np.zeros(n)
    A = Ab[:, :-1]
    b = Ab[:, -1]
    for i in range(n-1, -1, -1):
        x[i] = (b[i] - np.dot(A[i, i+1:], x[i+1:])) / A[i, i]
    return x
```
