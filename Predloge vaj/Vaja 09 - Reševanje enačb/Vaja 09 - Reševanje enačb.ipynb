### Bisekcijska metoda

```python
def bisekcija(fun, x1, x2, tol=1e-5, maxiter=50):
    """ Bisekcijska metoda numeričnega iskanja ničel. """
    if np.sign(fun(x1)) != np.sign(fun(x2)):
        n = int(np.ceil(np.log(np.abs(x2-x1)/tol) / np.log(2)))
        for i in range(min([n, maxiter])):
            x3 = (x1 + x2) / 2
            if np.sign(fun(x1)) != np.sign(fun(x3)):
                x2 = x3
            else:
                x1 = x3
        return x3
    else:
        raise Exception('Napačen začetni približek!')
```

### Sekantna metoda

```python
def sekantna(fun, x1, x2, tol=1e-5, maxiter=50):
    """ Sekentna metoda numeričnega iskanja ničel. """
    for i in range(maxiter):
        f1 = fun(x1)
        f2 = fun(x2)
        x3 = x2 + f2 * (x2 - x1) / (f1 - f2)
        x1 = x2
        x2 = x3
        if np.abs(x1 - x2) < tol:
            return x3
    return x3
```

### Newton-Raphsonova metoda

```python
def newton_raphson(fun, x0, dfun, tol=1e-5, maxiter=50):
    """ Newton-Raphson metoda numeričnega iskanja ničel. """
    for i in range(maxiter):
        x1 = x0 - fun(x0) / dfun(x0)
        if np.abs(x1 - x0) < tol:
            return x1
        x0 = x1
    return x0
```
