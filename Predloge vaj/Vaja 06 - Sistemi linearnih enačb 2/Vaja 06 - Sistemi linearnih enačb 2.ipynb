## Pivotiranje

#### Spomnimo se algoritma Gaussove eliminacije od zadnjič:

```python
def gauss_eliminacija(A, b):
    """Gauss eliminacija razširjene matrike [A|b]."""
    Ab = np.column_stack((A, b))
    
    for p, pivot_vrsta in enumerate(Ab[:-1]):
        for vrsta in Ab[p+1:]:
            if vrsta[p]:
                m = vrsta[p]/pivot_vrsta[p]
                vrsta[p:] = vrsta[p:] - pivot_vrsta[p:] * m
    return Ab
```

```python
def gauss_resitev(Ab):
    """Rešitev sistema enačb po gaussovi eliminaciji - obratno vstavljanje."""
    n = Ab.shape[0]
    x = np.zeros(n)
    A = Ab[:, :-1].copy()
    b = Ab[:, -1].copy()
    for i in range(n)[::-1]:
        x[i] = (b[i] - np.dot(A[i, i+1:], x[i+1:])) / A[i, i]
    return x
```

### Gaussova eliminacija z delnim pivotiranjem

Postopek je enak metodi Gaussove eliminacije, ki smo jo že spoznali, z dodatkom:
* Po vsaki izbiri pivotne vrstice ``i`` v prvi ``for`` zanki med vsemi vrsticami $j \geq p$ poiščemo indeks ``i_max`` tiste, ki ima v stolpcu ``i`` element z najvišjo absolutno vrednostjo.
* Če najprej izbrana vrstica ``i`` ni vrstica z največjo vrednostjo v stolpcu ``i``, vrstici zamenjamo med sabo z uporabo naprednega rezanja:

    ``Ab[[i], :], Ab[[i_max], :] = Ab[[i_max], :], Ab[[i], :]``
    
    
* Nadaljujemo postopek Gaussove eliminacije, kjer je nova vrsta ``i`` naša pivotna vrsta.

## Računanje inverza matrike

Pri računanju inverza matrike $\mathbf{A}$ gre v bistvu za reševanje sistema enačb:
$$\mathbf{AX} = \mathbf{I}$$

Iščemo torej tako matriko $\mathbf{X}$, ki pri množenju z $\mathbf{A}$ da enotsko matriko. Velja torej $\mathbf{X} = \mathbf{A}^{-1}$ Gre za reševanje več sistemov linearnih enačb, oblike:

$$\mathbf{A}\mathbf{x}_i = b_i$$

kjer sta $x_i$ in $b_i$ $i$-ta stolpca matrik $\mathbf{X}$ in $\mathbf{I}$ (iz definicije množenja matrik).

## Iterativne metode reševanja sistemov linearnih enačb

### Gauss-Seidel

```python
def gauss_seidel(A, b, x0, niter=10):
    """Gauss-Seidel iterativna metoda."""
    odstopki = []
    x = x0.copy()
    AA = A.copy()
    np.fill_diagonal(AA, 0)
    for i in range(niter):
        x_star = x.copy()
        for i in range(len(b)):
            x[i] = (b[i] - np.dot(AA[i], x)) / A[i,i]
        odstopki.append(np.linalg.norm(x_star - x))
    return x, np.array(odstopki)
```
