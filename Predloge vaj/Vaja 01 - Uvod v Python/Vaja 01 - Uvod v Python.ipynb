## While zanka

````python
# While zanka
import time
zacetek = time.time() # čas v sekundah
i = 0
while  time.time() - zacetek < 6:
    print('iteracija {:d}: '.format(i), time.time() - zacetek)
    time.sleep(1)
    i += 1
````
