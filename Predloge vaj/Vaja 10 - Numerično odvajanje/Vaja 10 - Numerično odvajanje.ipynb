$$\omega = 2\pi f \quad \text{in} \quad f=4~\text{Hz}$$

#### Izboljšanje približka odvoda z Richardsonovo ekstrapolacijo

Izboljšan približek numerične vrednosti odvoda $f'(x_i)$ pri koraku $h$ in $h/2$ (red napake metode $n$):

$$
\bar{f}'(x_i) = \frac{2^n\, f'_0\Big(x_i, \frac{h}{2}\Big) - f'_0(x_i, h)}{2^n-1}
$$
