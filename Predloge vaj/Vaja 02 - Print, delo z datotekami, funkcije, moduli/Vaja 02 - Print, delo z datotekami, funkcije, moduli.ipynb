## Funkcija print, formatiranje nizov

### Formatiranje števil

```python
# Izpis števil
st = 189.19341
formati = ['', 'g', 'f', '+f', '.2f', '3.2f', '8.2f', '<8.2f', '^8.2f', '9.0f', '.2e', '8.2e', '.5e']
print(f'Izpisujmo število {st}:')
print('format\t|izpis|')
print(20*'-')
for f in formati:
    print(f'{f!r:}\t|{st:{f:s}}|')
```

## Funkcije

```python
# Prva funkcija
def prva_funkcija(a, b, c=1):
    """
    Funkcija, ki sprejme vrednosti `a`, `b` in `c` in jih izpiše.
    Vrača vrednot `a*b*c`.
    """
    print(f'a = {a}')
    print(f'b = {b}')
    print(f'c = {c}')
    return a*b*c
```

## Delo z datotekami

```python
# Delo z datotekami
pot = 'datoteka.txt'
datoteka = open(pot, mode='w+')
print('Trenutna vsebina datoteke:')
print(datoteka.readlines(), '\n')

datoteka.write('Hello, file!\n')
print('Vrstica 2', file=datoteka)
print('Zapisovanje v datoteko...')

print('Nova vsebina datoteke:', end='\t')
datoteka.seek(0)
print(datoteka.readlines())
datoteka.close()
```

### Izpeljevanje seznamov, terk, slovarjev...

```python
# Izpeljevanje seznamov
[(k, podatki[k]) for k in podatki]
```

### Odstranjevanje, zamenjevanje znakov v nizu, deljenje nizov

```python
# Delo z nizi (split, replace...)
seznam_vrstic = [f'{k:s}\t{v:}\n' for k, v in podatki.items()]
s1 = [vrstica.rstrip() for vrstica in seznam_vrstic]
s2 = [vrstica.replace('\t', ' = ') for vrstica in s1]
s3 = [vrstica.split(' = ') for vrstica in s2]
```

### Modul pickle

```python
# Uporaba modula `pickle`
import pickle
data = list(range(10000))
# pisanje:
with open('ime.pkl', 'wb') as pklfile:
    pickle.dump(data, pklfile, protocol=-1)
# branje:
with open('ime.pkl', 'rb') as pklfile:
    seznam = pickle.load(pklfile)
```
