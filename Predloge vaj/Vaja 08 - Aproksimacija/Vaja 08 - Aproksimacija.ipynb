## Aproksimacija

##### Primer:

```python
N = 100
x = np.linspace(0, 1, N)
y = -3*x + 5 + 0.5*np.random.randn(len(x))

A = np.column_stack((x, np.ones_like(x)))
k, n = np.linalg.pinv(A).dot(y)

x_ = np.linspace(0, 1, 10*N)

plt.figure()
plt.plot(x, y, 'r.', label='točke')
plt.plot(x_, k*x_ + n, 'k', lw=3, label='aproksimacija', alpha=0.7)
plt.legend()
plt.show()
```
