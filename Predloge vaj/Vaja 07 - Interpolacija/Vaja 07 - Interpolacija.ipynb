## Interpolacija

### Interpolacija s polinomom

$n+1$ točk inteproliramo s polinomom stopnje $n$. Rešujemo sistem $n+1$ enačb oblike:

$$y_i = a_0 x^{n} + a_1 x^{n-1} + \dots + a_{n-1}x + a_{n}$$

### Interpolacija s polinomom - Lagrangeva metoda

Za vsako izmed interpolairanih točk $x_i$ zapišemo Lagrangev polinom $l_i$ stopnje $n$:

$$l_i(x) = \prod_{j=0, j\neq i}^{n} \frac{x-x_j}{x_i-x_j}$$

Lagrangevo interpolacijsko krivuljo nato izračunamo:

$$P_n(x) = \sum_{i=0}^{n}y_il_i(x)$$

## [video](https://www.youtube.com/watch?v=c_TbU8XBHsw)

```python
def lagrange_i(x, x_int, i):
    """ 
    Vrne vrednosti i-tega Lagrangevega polinoma
   
    x: seznam vrednosti x točk, ki jih interpoliramo
    x_int: neodvisna spremenljivka (točke, pri katerih iščemo vrednosti)
    i: indeks Lagrangevega polinoma
    """
    Li = 1.0
    for j in range(len(x)):
        if j != i:
            Li *= (x_int-x[j]) / (x[i]-x[j])
    return Li
```

```python
x = np.linspace(0,1,4)
y = np.random.rand(len(x))
x_int = np.linspace(np.min(x), np.max(x), 50)
plt.figure()
plt.axhline(y=0, color='k')
plt.axhline(y=1, color='k')
plt.plot(x, y, 'ro')
for i in range(len(x)):
    plt.axvline(x=x[i], color='k')
    plt.plot(x_int, lagrange_i(x, x_int, i), label=r'$l_{!s}$'.format(i))
plt.title('Lagrangevi polinomi')
plt.legend()
```
