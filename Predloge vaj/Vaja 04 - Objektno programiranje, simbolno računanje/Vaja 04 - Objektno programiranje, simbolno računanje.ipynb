## Osnove objektnega programiranja

```python
# Primer 1: razred Nakljucje
class Nakljucje:
    """Razred, ki izriše naključne točke."""

    def __init__(self, N=5, x_obmocje=(0,1), y_obmocje=(0,1), stil='ro'):
        """Konstruktor razreda."""
        self.N = N
        self.stil = stil
        self.xrange = x_obmocje
        self.yrange = y_obmocje

    def izris(self):
        """Metoda razreda, ki izračuna in izriše naključne točke."""
        self.X = np.linspace(self.xrange[0], self.xrange[1], self.N)
        self.Y = np.random.rand(self.N)*(self.yrange[1]-self.yrange[0]) + self.yrange[0]

        plt.plot(self.X, self.Y, self.stil)
```

Primer uporabe različnih možnosti pri konstruiranju objekta razreda ``Nakljucje``:

```python
objekt_2 = Nakljucje(50, (10, 100), (0,5), stil='c*')
objekt_2.izris()
```

Primer razreda, ki nadgradi funkcionalnost razreda Nakljucje tako, da ne potrebujemo funkcije izris, izriše koordinatno mrežo:

```python
# Primer 2: dedovanje razredov
class Nakljucje2(Nakljucje):
    """Razred, izpeljan iz razreda ``Nakljucje``, ki samodejno izriše točke."""

    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.izris()
        plt.grid()
```

```python
# Primer 3: pretvorba simbolnega izraza v num. funkcijo s sym.lambdify:
import sympy as sym
import numpy as np
import matplotlib.pyplot as plt

simbol_x = sym.symbols('x')
simbolni_izraz = sym.sin(simbol_x**2)
sym.plot(simbolni_izraz, (simbol_x, 0, 3), title='simbolno')
numpy_funkcija = sym.lambdify(simbol_x, simbolni_izraz, 'numpy')
x_num_polje = np.linspace(0, 3, 100)
plt.plot(x_num_polje, numpy_funkcija(x_num_polje), 'r.')
plt.title('numerično')
plt.show() # potrebno, kadar ne uporabimo %matplotlib inline
```
