### Kaj ocenjujemo pri individualnem projektu

Namen projekta je prikazati **ustrezno izbiro in uporabo numeričnih metod, komentirati njihovo delovanje in ovrednotiti dobljene rezultate**. Pri ocenjevanju gledamo štiri področja; najpomembnejše so numerične metode.

**1. Numerične metode** (vsako od šestih predpisanih poglavij ocenjujemo ločeno)

* *Pravilnost in ustreznost metode:*
    * problem je rešen **numerično**; zgolj simbolna rešitev (`sympy`) ali zapis enačb ne zadostuje,
    * metoda ustreza problemu (npr. predoločen sistem rešite po metodi najmanjših kvadratov, periodičnega signala ne aproksimirate s polinomom, diferencialno enačbo višjega reda pretvorite v sistem enačb 1. reda itd.),
    * če je to mogoče je rezultat **preverjen**: s primerjavo z analitično rešitvijo, drugo numerično metodo ali obstoječo funkcijo iz `numpy`/`scipy`,
    * uporaba obstoječih funkcij (`numpy`, `scipy`) je zaželena; lastna implementacija mora biti utemeljena in primerjana z obstoječo funkcijo.
* *Argumentacija in komentarji*, pri vsaki uporabljeni metodi pisno pojasnite:
    * **zakaj** ste izbrali to metodo in ne katere druge (npr. Newtonova metoda ali bisekcija? zakaj LU razcep? zakaj zlepki in ne polinom?, itd.),
    * **na kratko**: kakšne so predpostavke metode in kakšne so njene **omejitve** (natančnost, konvergenca, red metode, izbira koraka, robne točke pri odvajanju ...),
    * kaj rezultat pomeni za vaš fizikalni problem (**interpretacija**).
    * **Pomembno:** v seminarju **ne ponavljate izpeljav, teoretičnega ozadja metod** itd., če to ni bistveno za razumevanje ali izvajanje vaše kode. Teorija je predmet pisnega dela izpita.

**2. Poročilo in koda**

* *Poročilo:*
    * osnovni podatki (ime in priimek, vpisna številka, datum) in izjava o avtorstvu,
    * definicija naloge z opisom in **sliko** problema ter navedbo virov,
    * poglavja po **predlogi** (predlogo in **primer projekta** najdete v mapi [projekt/](https://github.com/jankoslavic/pynm/tree/main/projekt) v repozitoriju predmeta); iz naslovov je razvidno, kje je katera numerična vsebina,
    * razlaga v `markdown` celicah, enačbe zapisane z LaTeX-om,
    * rezultati prikazani grafično (`matplotlib`), z označenimi osmi in enotami.
* *Koda:*
    * poročilo se po `Restart` $\to$ `Run All` izvede **brez napak**, poti do datotek so relativne (vse potrebne datoteke so priložene, koda deluje kjerkoli, ne le na vašem računalniku),
    * koda je berljiva in v smiselnem vrstnem redu,
    * večkrat uporabljene funkcije so zbrane v **lastnem modulu** (`.py`), ki ga v poročilo uvozite,
    * funkcije imajo **docstringe** (opis, parametri, vrnjena vrednost) in smiselne komentarje,
    * dodatni paketi so navedeni z navodili za namestitev (npr. `requirements.txt`).

**3. Lasten odnos in kreativni dodatek**

* *Lasten odnos:* problem je smiselna celota in ne nabor nepovezanih nalog; numerične vsebine izhajajo iz problema; analiza je poglobljena (npr. vpliv parametrov, primerjava metod, ocena napake); rezultati so kritično ovrednoteni (ali so fizikalno smiselni?).
* *Kreativni dodatek:* vsebina, ki na predavanjih in vajah ni bila obravnavana (glejte seznam spodaj). Ocenjujemo kakovost izvedbe, ne le prisotnost.

**4. Testi kode in/ali uporabniški vmesnik**

Vsaj en test lastnih funkcij (npr. z `numpy.testing` ali `pytest`) ali preprost uporabniški vmesnik. Test mora biti v poročilu izveden in uspešen.

**Najpogostejše pomanjkljivosti** preteklih projektov, zaradi katerih so bile ocene nižje:
* metoda je uporabljena, ni pa pojasnjeno, zakaj ta, kako deluje in kako natančen je rezultat,
* poglavje je rešeno le simbolno,
* lastna implementacija metode brez utemeljitve in primerjave z obstoječo funkcijo,
* rezultati brez interpretacije,
* koda se zaradi napačnih poti ali manjkajočih paketov ne požene.

Kot **kreativni dodatek** se tretira programska vsebina, ki na predavanjih in vajah ni bila posebej obravnavana. Kot na primer:


* animacija rezultatov z uporabo matplotlib-a,

Pripravili smo tudi [video navodila za reševanje](https://youtu.be/wedX7AdORvY).

**Časovna omejitev reševanja domačih nalog**

Domače naloge imajo definiran čas *okna za reševanje*:
* Do odprtja okna za reševanje je možno **predčasno reševanje naloge brez časovne omejitve**.

* V oknu za reševanje je dovoljen čas reševanja naloge omejen. Časovna omejitev teče od trenutka, ko *znotraj okna za reševanje* nalogo prvič odprete.
* Po zaprtju okna se naloga zapre, reševanje **ni več možno**.

Svetujemo vam, da *čim večji del naloge rešite v času predčasnega reševanja, pred odprtjem okna za reševanje*.

**Uporaba Jokerja za premik obveznosti**

* Uporaba jokerja **ni možna, če ste nalogo že odprli**.
