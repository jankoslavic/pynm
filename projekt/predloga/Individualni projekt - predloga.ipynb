# *Navodila za uporabo te predloge*<span class="tocSkip"></span>

Predlogo individualnega projekta uporabite tako, da to datoteko ustrezno preimenujete, dopolnite s svojimi podatki ter pod pripravljenimi naslovi dodate svojo vsebino. Pred oddajo **izbrišite to poglavje z navodili**.

Če individualni projekt pišete v svoji datoteki, poskrbite, da bo ta ustrezno oblikovana, da vsebuje zahtevane elemente (vključno z izjavo o avtorstvu) in da so numerične vsebine jasno označene v naslovih poglavij poročila.

Primer izdelanega projekta najdete v mapi [projekt/primer/](https://github.com/jankoslavic/pynm/tree/main/projekt/primer) v repozitoriju predmeta.

# Definicija naloge 

Definicija naloge naj vsebuje kratek opis problema, sliko in navedbo virov (učbenik, članek, lastne meritve ...)! Navedite tudi, katere podatke uporabljate in od kod so.

Slika 3.22 iz vira Slavič 2014:
<img src="slika1.png" width=250>

Namen individualnega seminarja je prikazati **ustrezno izbiro in uporabo numeričnih metod, komentirati njihovo delovanje in ovrednotiti dobljene rezultate**. Temu primerna naj bo izbira fizikalnega problema, katerega reševnje naj zajema vsa predpisana poglavja:

* *simbolno reševanje*, 
* *sistemi linearnih enačb*, 
* *interpolacija ali aproksimacija*,
* *iskanje ničel*, 
* *integriranje ali odvajanje*, 
* *reševanje diferencialnih enačb*.  

**Struktura poročila / poglavij:** Vrstni red in naslovi poglavij lahko po potrebi spremenite, da bodo ustrezali vaši nalogi (glejte [primer projekta](https://github.com/jankoslavic/pynm/tree/main/projekt/primer)). Pri tem poskrbite, da bo iz naslovov poglavij še vedno **razvidno, kje v poročilu obravnavate katero od zgornjih vsebin**.

Pri vsaki uporabljeni numerični metodi **pisno pojasnite**, zakaj ste jo izbrali, kakšne so njene omejitve in kaj rezultat pomeni za vaš problem. Kjer je mogoče, rezultat **preverite** (analitična rešitev, druga metoda, obstoječa funkcija iz `numpy`/`scipy`).

Podrobnosti o tem, kaj pri projektu ocenjujemo, ter navodila za oddajo najdete v [študijskem in izpitnem redu](https://jankoslavic.github.io/pynm/notebooks/studijski-in-izpitni-red/).

**Koda:** večkrat uporabljene funkcije zberite v lastnem modulu (`.py`) z docstringi in ga v poročilo uvozite. Dodatne pakete navedite z navodili za namestitev (npr. `requirements.txt`). Priložite vsaj en test lastnih funkcij (npr. `numpy.testing`) ali preprost uporabniški vmesnik.
